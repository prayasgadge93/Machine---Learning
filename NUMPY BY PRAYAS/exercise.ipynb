```python
import numpy as np
data=np.array([
    [18,85,78],   #student1
    [19,92,88],   #student2
    [17,76,95],   #student3
    [18,65,70],   #student4
    [20,90,85]    #student5
])
```

```python
data.shape
```

```python
#find the average age of student
np.mean(data[:,0])
```

```python
#extract math marks of all student 
data[:,1]
```

```python
#find the highst science mark
np.max(data[:,2])
```

```python
#get detail of the student who scored more than 90 in math
data[data[:,1]>90]
```
